#### Q21. Create a CSV file
Create employees.csv containing employee ID, name, department, and salary. Use Python's csv module to write the header and five employee records.

```python
import csv
employees = [
    [101, "Aman", "IT", 45000],
    [102, "Priya", "HR", 35000],
    [103, "Rahul", "Finance", 50000],
    [104, "Neha", "Marketing", 40000],
    [105, "Vikas", "IT", 55000]
]
with open('employees.csv','w',newline='') as file:
    writer=csv.writer(file)
    writer.writerow(['employee_ID','Name','Department','Salary'])
    writer.writerows(employees)
print('employess.csv created successfully')
    
```

#### Q22. Read CSV data
Read employees.csv using both csv.reader() and csv.DictReader(). Print each employee's information.

```python

# Method 1: Using csv.reader()
print("Using csv.reader():")

with open("employees.csv", "r", newline="") as file:
    reader = csv.reader(file)

    header = next(reader)  # Skip the header

    for row in reader:
        print(
            f"ID: {row[0]}, Name: {row[1]}, "
            f"Department: {row[2]}, Salary: {row[3]}"
        )
```

```python
# Method 2: Using csv.DictReader()
print("\nUsing csv.DictReader():")

with open("employees.csv", "r", newline="") as file:
    reader = csv.DictReader(file)

    for employee in reader:
        print(
            f"ID: {employee['employee_ID']}, "
            f"Name: {employee['Name']}, "
            f"Department: {employee['Department']}, "
            f"Salary: {employee['Salary']}"
        )
```

#### Q26. Read JSON data
Create a JSON file containing product names, prices, and stock quantities. Read it using json.load() and calculate the total inventory value.

```python
import json

products = [
    {"name": "Laptop", "price": 50000, "stock": 10},
    {"name": "Mouse", "price": 500, "stock": 50},
    {"name": "Keyboard", "price": 1500, "stock": 20},
    {"name": "Monitor", "price": 12000, "stock": 5}
]

with open("products.json", "w") as file:
    json.dump(products, file, indent=4)

print("JSON file created successfully!")


with open("products.json", "r") as file:
    products = json.load(file)

total_value = 0

for product in products:
    value = product["price"] * product["stock"]
    total_value += value
    print(product["name"], ":", value)

print("Total inventory value:", total_value)
```

#### Q28. Understand JSON functions
Explain and demonstrate the difference between json.load(), json.loads(), json.dump(), and json.dumps().

```python
#json.dump()-writing into the file
customer = {
    "name": "Aman",
    "city": "Haldwani",
    "total_spent": 12500
}

with open("customer.json", "w") as file:
    json.dump(customer, file, indent=4)

print("Data saved successfully!")
```

```python
#json.load()-reading from the file
with open("customer.json", "r") as file:
    data = json.load(file)

print(data)
print(data["name"])
```

```python
#json.dumps() — convert Python data into a JSON string
customer = {"name": "Aman","city": "Haldwani","total_spent": 12500}
json_string = json.dumps(customer, indent=4)
print(json_string)
print(type(json_string))
```

```python
#json.loads() — convert a JSON string into Python data
json_string = '{"name": "Aman", "city": "Haldwani", "total_spent": 12500}'

data = json.loads(json_string)

print(data)
print(data["name"])
print(type(data))
```

#### Q29. Handle invalid JSON
Write a program that reads a JSON file and handles malformed JSON, missing keys, and invalid data types without crashing.

```python
players = [
    {
        "name": "Virat Kohli",
        "runs": 12000,
        "wickets": 4,
        "matches": 280
    },
    {
        "name": "Rohit Sharma",
        "runs": 10000,
        "wickets": 8,
        "matches": 260
    },
    {
        "name": "Jasprit Bumrah",
        "runs": 200,
        "wickets": 150,
        "matches": 100
    }
]

with open("cricket_stats.json", "w") as file:
    json.dump(players, file, indent=4)

print("Cricket JSON file created successfully!")
```

```python
import json

try:
    with open("cricket_stats.json", "r") as file:
        players = json.load(file)

    if not isinstance(players, list):
        raise ValueError("JSON data must contain a list of players.")

    for player in players:
        if not isinstance(player, dict):
            print("Invalid player record. Skipping.")
            continue

        required_keys = ["name", "runs", "wickets", "matches"]

        missing_keys = [
            key for key in required_keys if key not in player
        ]

        if missing_keys:
            print("Missing keys:", missing_keys)
            continue

        if not isinstance(player["name"], str):
            print("Invalid player name. Skipping.")
            continue

        if any(
            isinstance(player[key], bool)
            or not isinstance(player[key], int)
            or player[key] < 0
            for key in ["runs", "wickets", "matches"]
        ):
            print("Invalid statistics for:", player["name"])
            continue

        print("Player:", player["name"])
        print("Runs:", player["runs"])
        print("Wickets:", player["wickets"])
        print("Matches:", player["matches"])
        print("-" * 30)

except json.JSONDecodeError:
    print("Error: Cricket file contains malformed JSON.")

except FileNotFoundError:
    print("Error: Cricket file not found.")

except ValueError as e:
    print("Data error:", e)

except OSError as e:
    print("File error:", e)
```
