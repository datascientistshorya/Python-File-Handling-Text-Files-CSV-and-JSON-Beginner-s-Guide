#### Q4. Read a file line by line
Read sales.txt using read(), readline(), and readlines(). Explain the difference between their outputs.

```python
file=open('sales.txt','r')
content=file.read()
print(content,'\nThis reads the complete file\n')

file.seek(0)  # Move the pointer to the beginning
content1=file.readline()
print(content1,'\nThis reads a single line from the file\n')

file.seek(0)
content2=file.readlines()
print(content2,'\nThis reads lines from the file into a list')
```

#### Q8. Use a context manager
Read a file using with open(). Explain why this approach is preferred over manually opening and closing a file.

```python
with open('sales.txt', 'r') as file:
    content = file.read()
    print(content)
```

Explanation:
- with open() opens the file in read mode.
- file.read() reads the complete file.
- Once the with block finishes, Python automatically closes the file.
- This approach is preferred because it ensures the file is closed even if an exception occurs inside the block, helping prevent resource leaks.
- With manual closing, if an exception occurs before file.close(), the file might remain open until it is closed elsewhere or cleaned up.

#### Q9. Count file contents
Write a program to count the number of lines, words, and characters in a text file. Make sure the program also handles an empty file.

```python
with open('sales.txt','r') as file:
    content = file.read()

    if content == '':
        print("The file is empty.")
        print("Lines: 0")
        print("Words: 0")
        print("Characters: 0")
    else:
        lines = content.splitlines()
        words = content.split()
        characters = len(content)

        print("Lines:", len(lines))
        print("Words:", len(words))
        print("Characters:", characters)
        print(lines,words)
```

#### Q10. Mini-project: Expense tracker
Create a program that accepts five expenses, saves them to expenses.txt, reads the saved data, and calculates the total expenses.

```python
expenses=[
    {'food':12000},
    {'travel':3000},
    {'rent':18000},
    {'electricity':2700},
    {'internet':1000}]
with open('expenses.txt', 'w') as file:
    for i in expenses:
        for category, amount in i.items():
            file.write(f"{category}: {amount}\n")


total = 0

with open('expenses.txt', 'r') as file:
    for line in file:
        category, amount = line.strip().split(':')
        total += int(amount)

print("Total expenses:", total)

```
