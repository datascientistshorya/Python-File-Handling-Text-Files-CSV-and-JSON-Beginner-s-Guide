#### Q15. Search for a keyword
Search a text file for a given keyword. Print every matching line along with its line number

```python
keyword = input("Enter keyword to search: ")

with open('sales.txt', 'r') as file:
    found = False

    for line_number, line in enumerate(file, start=1):
        if keyword.lower() in line.lower():
            print(f"Line {line_number}: {line.strip()}")
            found = True

    if not found:
        print("Keyword not found.")
```

#### Q17. Handle file exceptions
Write a program that handles FileNotFoundError and PermissionError when attempting to read a file. Explain when each exception may occur.

```python
try:
    file=open('random.txt','r')
    print(file.read())
except FileNotFoundError as e:
    print('file or path doesnt exist',e)
```

```python
try:
    file=open('sample.txt','r')
    print(file.read())
except FileNotFoundError as e:
    print('file or path doesnt exist',e)
except PermissionError as e:
    print('Permission Denied to Access the file!!')
```

Explanation

1. FileNotFoundError

This exception occurs when Python attempts to open a file that does not exist at the specified path.

Example: If sample.txt does not exist in the current working directory, Python raises FileNotFoundError.

Solution: Check whether the filename and file path are correct.

2. PermissionError

This exception occurs when the operating system denies Python permission to read the file.

Example: The file exists, but the current user does not have the required read permissions.

Solution: Check the file's permissions or request access from the file owner or system administrator.

#### Q18. Create a log file
Create a logging program that appends messages to application.log without overwriting previous messages. Include a timestamp with each entry.

```python
from datetime import datetime

message = input("Enter log message: ")
timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

with open("application.log", "a", encoding="utf-8") as file:
    file.write(f"[{timestamp}] {message}\n")

print("Log entry added successfully.")
```

#### Q19. Combine multiple text files
Read all .txt files from a specified directory and combine their contents into combined.txt. Ensure the output file is not accidentally read as an input.

```python
from pathlib import Path

directory = Path(".")
output_file = directory / "combined.txt"

with open(output_file, "w", encoding="utf-8") as output:
    for file_path in directory.glob("*.txt"):
        if file_path != output_file:
            with open(file_path, "r", encoding="utf-8") as file:
                output.write(file.read())
                output.write("\n")

print("All text files combined successfully.")
```

#### Q20. Mini-project: Text analyzer
Build a program that analyzes a text file and reports the total lines, words, characters, most frequent words, and occurrences of a user-provided keyword. Handle missing and empty files.

```python
try:
    key = input("Enter the word to be searched: ")
    count = 0

    with open("combined.txt", "r", encoding="utf-8") as file:
        content = file.read()

    lines = content.splitlines()
    words = content.split()
    characters = len(content)

    freq = {}

    for word in words:
        word = word.strip(".,!?").lower()

        if word:
            freq[word] = freq.get(word, 0) + 1

    key = key.strip(".,!?").lower()
    count = freq.get(key, 0)

    print("Total lines:", len(lines))
    print("Total words:", len(words))
    print("Total characters:", characters)
    print(key, "occurred", count, "times in the file")

    if freq:
        most_frequent = max(freq, key=freq.get)
        print("Most frequent word:", most_frequent)
        print("Frequency:", freq[most_frequent])
    else:
        print("The file is empty or contains no words.")

except FileNotFoundError:
    print("Error: combined.txt was not found.")

except PermissionError:
    print("Error: You do not have permission to read the file.")
```
